```sql
                            Replicate, Space, Patindex, Replace and Stuff functions 

Used Table: tblEmployee6

-- ### 1. REPLICATE() FINCTION                                    
1 REPLICATE(String_To_Be_Replicated, Number_Of_Times_To_Replicate) - Repeats the given string, for the specified number of times.

Example: SELECT REPLICATE('Pragim', 3)
Example Databricks: SELECT REPEAT ('Pragim', 3)
Output: Pragim Pragim Pragim 

        -- A practical example of using REPLICATE() function: Let's mask the email with 5 * (star) symbols. The output should be as shown below.
| FirstName | LastName | Email           |
| --------- | -------- | --------------- |
| Sam       | Sony     | Sa*****@aaa.com |
| Ram       | Barber   | Ra*****@aaa.com |
| Sara      | Sanosky  | Sa*****@ccc.com |
| Todd      | Gartner  | To*****@bbb.com |
| John      | Grover   | Jo*****@aaa.com |
| Sana      | Lenin    | Sa*****@ccc.com |
| James     | Bond     | Ja*****@bbb.com |
| Rob       | Hunter   | Ro*****@ccc.com |
| Steve     | Wilson   | St*****@aaa.com |
| Pam       | Broker   | Pa*****@bbb.com |

select * from tblEmployee6
Select FirstName, LastName, SUBSTRING(Email, 1, 2) + REPLICATE('*',5) + 
SUBSTRING(Email, CHARINDEX('@',Email), LEN(Email) - CHARINDEX('@',Email)+1) as Email
from tblEmployee6
-- OR
select FirstName, LastName, Email, CONCAT(SUBSTRING(Email, 1, 2), REPEAT('*', 5), SUBSTRING(Email,INSTR(Email,'@'))) from tblEmployee6 

-- ### 2. SPACE() FINCTION   
SPACE(Number_Of_Spaces) - Returns number of spaces, specified by the Number_Of_Spaces argument.
select space(2)

Example: Inserts 5 spaces between FirstName and LastName ?
Select FirstName + SPACE(5) + LastName as FullName
From tblEmployee
-- OR
select FirstName , LastName, CONCAT(FirstName, SPACE(5), Lastname) as Fullname 
from tblEmployee6

-- ### 3. PATINDEX() FINCTION  
ATINDEX('%Pattern%', Expression)
Returns the starting position of the first occurrence of a pattern in a specified expression. It takes two arguments, the pattern to be searched and the expression. PATINDEX() is simial to CHARINDEX(). With CHARINDEX() we cannot use wildcards, where as PATINDEX() provides this capability. If the specified pattern is not found, PATINDEX() returns ZERO.

Example: 
Select Email, PATINDEX('%@aaa.com', Email) as FirstOccurence 
from tblEmployee6
Where PATINDEX('%@aaa.com', Email) > 0
-- OR
SELECT Email
FROM tblEmployee6
WHERE Email RLIKE '@aaa.com'

-- ### 4. REPLACE() FINCTION  
REPLACE(String_Expression, Pattern , Replacement_Value)
Replaces all occurrences of a specified string value with another string value.

Example: All .COM strings are replaced with .NET
Select Email, REPLACE(Email, '.com', '.net') as ConvertedEmail
from  tblEmployee6

-- ### 5. STUFF() FINCTION  
STUFF(Original_Expression, Start, Length, Replacement_expression)
STUFF() function inserts Replacement_expression, at the start position specified, along with removing the charactes specified using Length parameter.

Example:
Select FirstName, LastName,Email, STUFF(Email, 2, 3, '*****') as StuffedEmail
From tblEmployee6
2: start position
3: next 3 character starting from 3
*****: 5 * will be applied
-- OR (to visualise code break part by part in concat) Databricks 
-- select SUBSTRING('sam.sony@aaa.com', 5) -> sony@aaa.com
SELECT FirstName, LastName,Email, concat(SUBSTRING(Email, 1, 1), '*****', SUBSTRING(Email,5) ) 
from tblEmployee6



```
