```sql
                                      -- JOIN ---
Joins in SQL server are used to query (retrieve) data from 2 or more related tables. In general tables are related to each other using foreign key constraints.

-- In SQL server, there are different types of JOINS.
1. CROSS JOIN
2. INNER JOIN 
3. OUTER JOIN 

-- Outer Joins are again divided into 3 types
1. Left Join or Left Outer Join
2. Right Join or Right Outer Join
3. Full Join or Full Outer Join

Used table: tblEmployee2 (10 rows), tblDepartment2 (4 rows)

| **Join Type**  | **Purpose**                                                             |
| -------------- | ----------------------------------------------------------------------- |
| **Cross Join** | Returns Cartesian product of the tables involved in the join.           |
| **Inner Join** | Returns only the matching rows. Non-matching rows are eliminated.       |
| **Left Join**  | Returns all the matching rows + non-matching rows from the left table.  |
| **Right Join** | Returns all the matching rows + non-matching rows from the right table. |
| **Full Join**  | Returns all rows from both tables, including the non-matching rows.     |

-- General Formula for Joins
SELECT ColumnList
FROM LeftTableName
JOIN_TYPE RightTableName
ON JoinCondition

                                      -- JOIN / INNER JOIN --
SELECT Name, Gender, Salary, DepartmentName
FROM tblEmployee2
INNER JOIN tblDepartment2
ON tblEmployee2.DepartmentId = tblDepartment2.Id

-- OR

SELECT Name, Gender, Salary, DepartmentName
FROM tblEmployee2
JOIN tblDepartment2
ON tblEmployee2.DepartmentId = tblDepartment2.Id

                                  -- LEFT JOIN / LEFT OUTER JOIN --
SELECT Name, Gender, Salary, DepartmentName
FROM tblEmployee2
LEFT OUTER JOIN tblDepartment2
ON tblEmployee2.DepartmentId = tblDepartment2.Id

OR

SELECT Name, Gender, Salary, DepartmentName
FROM tblEmployee2
LEFT JOIN tblDepartment2
ON tblEmployee2.DepartmentId = tblDepartment2.Id
                                      -- RIGHT JOIN --
SELECT Name, Gender, Salary, DepartmentName
FROM tblEmployee2
RIGHT OUTER JOIN tblDepartment2
ON tblEmployee2.DepartmentId = tblDepartment2.Id

OR

SELECT Name, Gender, Salary, DepartmentName
FROM tblEmployee2
RIGHT JOIN tblDepartment2
ON tblEmployee2.DepartmentId = tblDepartment2.Id      
                                      -- FULL JOIN --                                      
SELECT Name, Gender, Salary, DepartmentName
FROM tblEmployee2
FULL OUTER JOIN tblDepartment2
ON tblEmployee2.DepartmentId = tblDepartment2.Id

OR

SELECT Name, Gender, Salary, DepartmentName
FROM tblEmployee2
FULL JOIN tblDepartment2
ON tblEmployee2.DepartmentId = tblDepartment2.Id   
                                    -- CROSS JOIN --
Cross Join shouldn't have ON clause. 10*4 = 40 rows will be there
SELECT Name, Gender, Salary, DepartmentName
FROM tblEmployee2
CROSS JOIN tblDepartment2    

OR

SELECT Name, Gender, Salary, DepartmentName
FROM tblEmployee2
CROSS JOIN tblDepartment2        
ORDER BY Name                            
```
