```sql
                                  -- Advance JOIN or Intellegent Join --
1. Retrieve only the non matching rows from the left table
2. Retrieve only the non matching rows from the right table
3. Retrieve only the non matching rows from both the left and right table (FULL JOIN)

Used table: tblemployee2, tbldepartment2
select * from tblEmployee2;
select * from tblDepartment2;

                          -- Retrieve rows from Left Join (8 rows matching 2 non matching, so 2 nonmatching rows have to find out)
SELECT Name, Gender, Salary, DepartmentName
FROM           tblEmployee2 E
LEFT JOIN   tblDepartment2 D
ON                 E.DepartmentId = D.Id
WHERE        D.Id IS NULL
-- OR Using LEFT ANTI JOIN. [can also use RIGHT ANTI JOIN]
SELECT Name, Gender, Salary, DepartmentId
FROM tblEmployee2 E
LEFT ANTI JOIN tblDepartment2 D
ON E.DepartmentId = D.Id

OR

SELECT Name, Gender, Salary, DepartmentName
FROM           tblEmployee2 E
LEFT JOIN   tblDepartment2 D
ON                 E.DepartmentId = D.Id
WHERE        E.DepartmentId IS NULL

                          -- Retrieve rows from Right Join 
SELECT         Name, Gender, Salary, DepartmentName
FROM             tblEmployee2 E
RIGHT JOIN    tblDepartment2 D
ON                   E.DepartmentId = D.Id
WHERE          E.DepartmentId IS NULL

                          -- Retrieve rows from Full Join Both left & right table
SELECT         Name, Gender, Salary, DepartmentName
FROM              tblEmployee2 E
FULL JOIN      tblDepartment2 D
ON                   E.DepartmentId = D.Id
WHERE          E.DepartmentId IS NULL
OR                   D.Id IS NULL
```
