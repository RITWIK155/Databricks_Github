```sql
                      -- INTERVIEW
-- 1. Select specific or all columns
-- 2. Distinct rows
-- 3. Filtering with where clause.
-- 4. Wild Cards in SQL Server
-- 5. Joining multiple conditions using AND and OR operators
-- 6. Sorting rows using order by
-- 7. Selecting top n or top n percentage of rows

-- select ID,Name,Email, GenderId, city FROM [sample].[dbo].[tblPerson]
-- sample: Database
-- dbo: Schema
-- tblPerson: Table

CREATE TABLE workspace.default.tblperson2 (
  ID INT NOT NULL,
  Name STRING,
  Email STRING,
  GenderId INT ,
  Age INT,
  City STRING)

INSERT into tblperson2(ID,Name,Email,GenderId,Age,City) values 
(1,'TOM','t@t.com',1,23,'London'),
(2,'John','J@J.com',1,20,'New York'),
(3,'Mary','mary.mary.com',2,21,'Sydney'),
(4,'Josh','josh.j@dell.com',1,29,'London'),
(5,'Sara','sara@abc.com',2,25,'Mumbai')
                  -- INTERVIEW ANSWER
-- 1
select * from tblperson2
-- 2  (Distinct city & name of the person)
select distinct City from tblperson2
select distinct City ,Name from tblperson2
-- 3 (Peoples who are residing in London)
select * from tblperson2 where city = 'London'
-- 4 (Wild Card & Characters - Khata)
select * from tblperson2 where age in (20,23,29)
select * from tblperson2 where age between 20 and 25
-- (Person who are resifing in a city starting from L)
select * from tblperson2 where city like 'L%'
-- (Find email valid or not)
select * from tblperson2 where email like '%@%'
select * from tblperson2 where email NOT like '%@%'
-- (Find persons whose email is one character before @ and after @ Ex: a@a.com, b@b.com)
select * from tblperson2 where email like '_@_.com'
-- (Name all the people whose name start with M, S, T)
-- SQL SERVER
select * from tblperson2 where Name like '[MST]%'
select * from tblperson2 where Name like '[^MST]%'  --NOT start with M,S,T
-- DATABRICKS regex matching
SELECT * FROM tblperson2 WHERE Name RLIKE '^(M|S|T)';
-- 5 (People living in London or Mumbai and age >= 25)
select * from tblperson2 where (city = 'London' or city = 'Mumbai') and age >= '25'
-- 6 use orderby to sort Name  alphabet ASC/DESC order
-- By default it is ASENDING
select * from tblperson2 order by Name
select * from tblperson2 order by Name DESC
select * from tblperson2 order by Age DESC
-- Sort by name ASC and age DESC. You cannot use AND in the ORDER BY clause.
select * from tblperson2 order by Name ASC , age DESC
-- 7 TOP keyword used when table is large 
-- select top 2/10 records from the table
-- SQL SERVER used, Databricks used limit
select top 2 * from tblperson2
-- Databricks
select * from tblperson2 limit 2
select Name, Age from tblperson2 limit 2
-- top 1% from the records
select top 1 % from tblperson2

https://www.youtube.com/watch?v=nYmoQ4r0DVw
```
