```sql
COALESCE(): COALESCE() returns the first Non NULL value from rows.

Used table: tblEmployee4

-- ### tblEmployee4
Id  |  FirstName  |  MiddleName  |  LastName
----+-------------+--------------+----------
1   |  Sam        |  NULL        |  NULL    
2   |  NULL       |  Todd        |  Tanzan  
3   |  NULL       |  NULL        |  Sara    
4   |  Ben        |  Parker      |  NULL    
5   |  James      |  Nick        |  Nancy   

-- ### OUTPUT
Id  |  Name 
----+-------
1   |  Sam  
2   |  Todd 
3   |  Sara 
4   |  Ben  
5   |  James

select ID, COALESCE(FirstName, MiddleName, LastName) as name from tblEmployee4
```
