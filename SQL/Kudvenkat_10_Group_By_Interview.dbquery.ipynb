```sql
In SQL Server we have got lot of aggregate functions. Examples
1. Count()
2. Sum()
3. avg()
4. Min()
5. Max()

Used table: tblEmployee

Group by clause is used to group a selected set of rows into a set of summary rows by the values of one or more columns or expressions. It is always used in conjunction with one or more aggregate functions.

-- I want an sql query, which gives total salaries paid by City. The output should be as shown below. Without Groupby city will not be counted will throw error.
Note: If you omit, the group by clause and try to execute the query, you get an error , Although "select City, SUM(Salary) from tblEmployee" it make sence but it is not correct becase how ssms not know to make groupby whom? so error occur - Column 'tblEmployee.City' is invalid in the select list because it is not contained in either an aggregate function or the GROUP BY clause. Because you cant apply Aggregate function in String column , so int only.
-- city | total_salary
-- London | 27300
-- New Yourk | 15500
-- Sydney | 7600
select * from tblEmployee
select *, sum(Salary) from tblEmployee GROUP BY City -- Error ,Because: see beneath
select City, SUM(Salary) from tblEmployee GROUP BY City 

-- Now, I want an sql query, which gives total salaries by City, by gender. The output should be as shown below.
select City, Gender, SUM(Salary) as TotalSalary from tblEmployee GROUP BY City ,Gender
select City, Gender, SUM(Salary) as TotalSalary from tblEmployee GROUP BY City ,Gender order by City

-- Total employee of tblEmployee table
select count(*) from tblEmployee
select count(ID) from tblEmployee

-- Query for retrieving total salaries and total number of employees by City, and by gender
select City, Gender, SUM(Salary) as TotalSalary, COUNT(ID) as TotalEmployee from tblEmployee GROUP BY City ,Gender 
-- if you want alias having blank space use [] SQL SERVER, Databricks ``
select City, Gender, SUM(Salary) as TotalSalary, COUNT(ID) as `Total Employee` from tblEmployee GROUP BY City ,Gender 

Filtering Groups:
-- Now I want only Male record with above query (WHERE clause)
select City, Gender, SUM(Salary) as TotalSalary, COUNT(ID) as TotalEmployee from tblEmployee
where Gender = 'Male'
GROUP BY City ,Gender 
--#1
select dept, avg(salary) 
from employees
where salary > 40000
group by dept
-- (HAVING clause)
select City, Gender, SUM(Salary) as TotalSalary, COUNT(ID) as TotalEmployee from tblEmployee
GROUP BY City ,Gender 
HAVING Gender = 'Male'
--#2
select dept, avg(salary)
from employees
HAVING avg(salary)>4500
-- WHERE & HAVING difference (VVI Common Interview Quistion)
1. When you use Where clause from the table only Male records are retrieved and then they are gouped.
2. When you using Having clause, all the rows from the table tblEmployee are retrieved and grouped by all gender and then only Male groups are shown so filtered out Female group.
3. WHERE filters rows before aggregation (GROUPING), where as, HAVING filters groups, after the aggregations are performed.
4. WHERE clause can be used with - Select, Insert, and Update statements. But HAVING clause can only be used with the Select statement.
5. Aggregate functions cannot be used in the WHERE clause, unless it is in a sub query contained in a HAVING clause, whereas, aggregate functions can be used in Having clause.
Ex: select * from tblEmployee where SUM(Salary) > 4000.  -- Will give error
select City, Gender, SUM(Salary) as TotalSalary, COUNT(ID) as TotalEmployee from tblEmployee
GROUP BY City ,Gender 
Having SUM(Salary)>4000
---------------
-- Another Example : https://www.youtube.com/shorts/PbQ71eSSHnw
| dept | salary |
|------|--------|
| IT   | 50000  |
| IT   | 60000  |
| HR   | 40000  |
| HR   | 30000  |  <-- crossed out

=
+--------+--------+
|  dept  | salary |
+--------+--------+
|   IT   |  50000 |   \
                      } group 1
|   IT   |  60000 |   /

|   HR   |  40000 |   \
                      } group 2
| --HR---|--30000-|   /  <-- (Crossed out)
+--------+--------+

select dept, avg(salary) 
from employees
where salary > 40000
group by dept
-- where
select dept, avg(salary)
from employees
where avg(salary)>4500
--having
select dept, avg(salary)
from employees
HAVING avg(salary)>4500

-- Where vs Having
-- WHERE
1) Where works on raw data .
2) Means it works before creating group.
3) where is used in select, update & delete
-- HAVING
1) Having works on the results produced by aggregating functions like sum(), count(), avg()
2) Having filter groups after aggregation(group)
3) But having only used with select clause

-------------------------------------------------------------------------------------------
                        Group By Importent Concept

-- ERROR #1: 
select EmpName, Gender, count(*) as total from tblEmployee10 GROUP BY DepartmentId -- Error [MISSING_AGGREGATION] The non-aggregating expression "EmpName" is based on columns which are not participating in the GROUP BY clause.
Error because: Chatgpt: When you use GROUP BY, every column in the SELECT list must either: Be inside an aggregate function (COUNT, SUM, etc.), or Be included in the GROUP BY clause. Because: 🟢 EmpName and Gender are not aggregated, 🟢 They are not in GROUP BY, 🟢 So SQL Server will give an error.
-- CORRECT
1️⃣ If you want employee count per department:
SELECT DepartmentId, COUNT(*) AS Total
FROM tblEmployee10
GROUP BY DepartmentId
2️⃣ If you want count by Gender inside department:
SELECT DepartmentId, Gender, COUNT(*) AS Total
FROM tblEmployee10
GROUP BY DepartmentId, Gender
3️⃣ If you want count per employee name:
SELECT EmpName, Gender, COUNT(*) AS Total
FROM tblEmployee10
GROUP BY EmpName, Gender

-- copilot ans: 1. You are grouping by DepartmentId, but selecting EmpName and Gender which are not part of the GROUP BY clause. 2. SQL requires that every column in the SELECT list (except aggregates like COUNT) must either be in the GROUP BY clause or wrapped in an aggregate function.

                    -- Another Example(SEE THIS)
select * from tblEmployee
select City, SUM(Salary) from tblEmployee GROUP BY City -- working fine

select *, sum(Salary) from tblEmployee GROUP BY City -- Error msg The non-aggregating expression "ID" is based on columns which are not participating in the GROUP BY clause. Because: When you use GROUP BY, every column in the SELECT list must either be included in the GROUP BY clause or wrapped in an aggregate function. And The columns other than City (like ID, Name, Gender, Salary) are not aggregated and not included in the GROUP BY clause, so Databricks SQL throws a [MISSING_AGGREGATION] error 
-- If you want to include other columns and do not care which value is shown for each group, you can use any_value() for those columns: select City, any_value(Name), SUM(Salary) from tblEmployee GROUP BY City




```
